```python
import numpy as np
from sklearn.linear_model import ElasticNet
```

```python
np.random.seed(42)
m = 100
X = 6 * np.random.randn(m,1) - 3
y = 0.5 * X ** 2 + X + 2 + np.random.randn(m,1)
```

```python
elastic_net = ElasticNet(alpha = 0.1, l1_ratio = 0.5)
elastic_net.fit(X,y)
elastic_net.predict([[1.5]])
```
