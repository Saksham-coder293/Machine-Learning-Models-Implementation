```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris 
iris = load_iris(as_frame = True)
list(iris)
```

```python
iris.data.head(3)
```

```python
iris.target_names
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
X = iris.data[["petal width (cm)"]]
y = iris.target_names[iris.target] == 'virginica'
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state =  42)
```

```python
log_reg = LogisticRegression(random_state = 42)
log_reg.fit(X,y)
```

```python
X_new = np.linspace(0,3,1000).reshape(-1,1)   #reshape to column vector 
y_proba = log_reg.predict(X_new)
y_proba = log_reg.predict_proba(X_new)
decision_boundary = X_new[y_proba[:, 1] >= 0.5][0, 0]
```

```python
plt.plot(X_new, y_proba[:, 0], "b--",linewidth = 2, label = "Not iris virginica proba")
plt.plot([decision_boundary, decision_boundary], [0, 1], "k:", linewidth=2, label="Decision_boundary")
plt.show()
```
